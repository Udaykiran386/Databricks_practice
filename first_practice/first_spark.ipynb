```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType,StructField, StringType, IntegerType, FloatType

schema = StructType([
    StructField("id", IntegerType(), True),
    StructField("name", StringType(), True),
    StructField("age", IntegerType(), True),
    StructField("salary", FloatType(), True),
])
# 1.Direct empty data frame with structType schema
df1 = spark.createDataFrame([], schema)
df1.printSchema()
display(df1)

```

```python
# 2. Create Empty dataframe using sql ddl string schema
ddl_schema = "firstname STRING,middlename STRING,lastname STRING"
df_ddl = spark.createDataFrame([],ddl_schema)
df_ddl.printSchema()
```

```python
# 3. Create Empty DataFrame Without Schema (0 columns, 0 rows)
df_no_schema = spark.createDataFrame([], StructType([]))
df_no_schema.printSchema()
```

```python
import pandas as pd
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType,StructField, StringType, IntegerType, FloatType

schema = StructType([
    StructField("id", IntegerType(), True),
    StructField("name", StringType(), True),
    StructField("age", IntegerType(), True),
    StructField("salary", FloatType(), True),
])
# 4. Create Empty DataFrame from an Empty Pandas DataFrame
empty_pd_df = pd.DataFrame(columns=["firstname", "middlename", "lastname"])
df_from_pandas = spark.createDataFrame(empty_pd_df, schema=schema)
df_from_pandas.printSchema()
```

```python
dept = [("Finance", 10), ("Marketing", 20), ("Sales", 30), ("IT", 40)]
df = spark.createDataFrame(data=dept, schema=["dept_name", "dept_id"])
display(df)

```

```python
dept_schema = StructType([
    StructField("dept_name", StringType(), True),
    StructField("dept_id", IntegerType(), True),
])
df_serverless_2 = spark.createDataFrame(dept, dept_schema)
df_serverless_2.printSchema()
display(df_serverless_2)
df_serverless_2.show()
```
